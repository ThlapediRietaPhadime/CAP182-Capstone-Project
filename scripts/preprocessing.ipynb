```python
import pandas as pd
import os
```

```python
df = pd.read_csv("bank-additional-full.csv" , sep=";")
```

```python
df.columns = [c.strip().lower().replace(".","_") for c in df.columns]
```

```python
df.shape
```

```python
df.head()
```

```python
df = df.rename(columns={"y": "deposit"})
df["deposit"] = df["deposit"].map({"yes": 1, "no": 0}).astype(int)
df["deposit"].value_counts()
```

```python
df["deposit"].value_counts(normalize=True)
```

```python
df = df.drop(columns=["duration"])
df.shape
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates().reset_index(drop=True)
df.shape
```

```python
test_size = 0.2
n_test = int(len(df) * test_size)
split_idx = len(df) - n_test

train_df = df.iloc[:split_idx].reset_index(drop=True)
test_df = df.iloc[split_idx:].reset_index(drop=True)

print(train_df.shape, test_df.shape)
```

```python
print("Train positive rate:", train_df["deposit"].mean())
print("Test positive rate:", test_df["deposit"].mean())
```

```python
SAVE_DIR = "data/processed"
os.makedirs(SAVE_DIR, exist_ok=True)

train_df.to_csv(os.path.join(SAVE_DIR, "train_clean.csv"), index=False)
test_df.to_csv(os.path.join(SAVE_DIR, "test_clean.csv"), index=False)
print("Saved")
```

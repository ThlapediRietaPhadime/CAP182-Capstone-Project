```python
import pandas as pd
import os
```

```python
train_df = pd.read_csv("data/processed/train_clean.csv")
test_df = pd.read_csv("data/processed/test_clean.csv")

print(train_df.shape, test_df.shape)
train_df.head()
```

```python
train_df["previously_contacted"] = (train_df["pdays"] != 999).astype(int)
test_df["previously_contacted"] = (test_df["pdays"] != 999).astype(int)

train_df["previously_contacted"].value_counts()
```

```python
train_df["contacted_before"] = (train_df["previous"] > 0).astype(int)
test_df["contacted_before"] = (test_df["previous"] > 0).astype(int)

train_df["contacted_before"].value_counts()
```

```python
categorical_columns = [
    "job", "marital", "education", "default", "housing", "loan",
    "contact", "month", "day_of_week", "poutcome"
]
```

```python
train_encoded = pd.get_dummies(train_df, columns=categorical_columns)
test_encoded = pd.get_dummies(test_df, columns=categorical_columns)

print(train_encoded.shape, test_encoded.shape)
```

```python
train_cols = train_encoded.columns
test_encoded = test_encoded.reindex(columns=train_cols, fill_value=0)

print(train_encoded.shape, test_encoded.shape)
```

```python
X_train = train_encoded.drop(columns=["deposit"])
y_train = train_encoded["deposit"]

X_test = test_encoded.drop(columns=["deposit"])
y_test = test_encoded["deposit"]

print(X_train.shape, X_test.shape)
```

```python
SAVE_DIR = "data/processed"
os.makedirs(SAVE_DIR, exist_ok=True)

X_train.to_csv(os.path.join(SAVE_DIR, "X_train.csv"), index=False)
X_test.to_csv(os.path.join(SAVE_DIR, "X_test.csv"), index=False)
y_train.to_csv(os.path.join(SAVE_DIR, "y_train.csv"), index=False)
y_test.to_csv(os.path.join(SAVE_DIR, "y_test.csv"), index=False)

print("Saved")
```
